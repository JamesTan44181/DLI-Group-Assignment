### **Step 3: Run the Training Notebook**

This is the main part of the script. It uses a library called `nbformat` to read the notebook file (`DLI_GROUP_Model.ipynb`) from the cloned repository. It then executes each code cell in order.

Crucially, the final cell is modified on the fly to ensure the trained model (`phishing_model.keras`) and test data are saved to Google Drive, not the temporary Colab storage.

```python
# Import necessary libraries to read and execute the notebook
import nbformat
from IPython import get_ipython
import pandas as pd # Ensure pandas is imported for the save step
import os # Import os for path joining

print("\n--> Step 3: Installing required library and executing the notebook...")
# Install the library needed to parse the notebook file
!pip install -q nbformat
print("Library installed.")

# --- Configuration ---
# Path to the notebook file within the cloned repository
notebook_path = '/content/DLI-Group-Assignment/DLI_GROUP_Model.ipynb'

# Define the save paths for the output files in the Google Drive Colab Notebooks directory
drive_dir = '/content/drive/MyDrive/Colab Notebooks/'
drive_model_path = os.path.join(drive_dir, 'phishing_model.keras')
drive_xtest_path = os.path.join(drive_dir, 'X_test.csv')
drive_ytest_path = os.path.join(drive_dir, 'y_test.csv')

# Ensure the Google Drive directory exists
os.makedirs(drive_dir, exist_ok=True)
# --- End of Configuration ---

try:
    with open(notebook_path) as f:
        nb = nbformat.read(f, as_version=4)

    # Get the IPython shell instance to run the code
    ipython_shell = get_ipython()

    # Execute each code cell from the notebook sequentially
    for i, cell in enumerate(nb.cells):
        if cell.cell_type == 'code':
            print(f"--- Running Cell {i+1} ---")
            code = cell.source

            # On the final cell, modify the save paths to point to the Google Drive directory
            if 'model.save("phishing_model.keras")' in code:
                print("Redirecting save paths to your Google Drive Colab Notebooks folder...")
                code = code.replace('model.save("phishing_model.keras")', f'model.save("{drive_model_path}")')
                code = code.replace('pd.DataFrame(X_test).to_csv("X_test.csv", index=False)', f'pd.DataFrame(X_test).to_csv("{drive_xtest_path}", index=False)')
                code = code.replace('pd.DataFrame(y_test).to_csv("y_test.csv", index=False)', f'pd.DataFrame(y_test).to_csv("{drive_ytest_path}", index=False)')

            # Execute the code in the current environment
            result = ipython_shell.run_cell(code)

            # Stop execution if any cell has an error
            if result.error_in_exec:
                print(f"!!! ERROR in Cell {i+1}: Execution stopped. !!!")
                break
    print("------------------------")

except FileNotFoundError:
    print(f"Error: The notebook file was not found at {notebook_path}")
except Exception as e:
    print(f"An error occurred: {e}")

print("\n--> Process Finished!")
print(f"✅ All steps completed. The trained model and test data should now be in your Google Drive Colab Notebooks folder.")
```

### **Evaluation: Performance Visualization**

This script generates the required evaluation plots:  
- **Confusion Matrix** for a clear summary of prediction accuracy  
- **ROC Curve** to assess class separation  
- **Precision-Recall (PR) Curve** to analyze the trade-off between precision and recall  


```python
# Import necessary libraries
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, precision_recall_curve, auc

# This script assumes 'y_test', 'y_pred', 'y_pred_prob', and 'roc_auc' are pre-defined

print("\n📈=============== Performance Visualization ===============📈")

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=['Benign', 'Phishing'], yticklabels=['Benign', 'Phishing'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()

# Precision-Recall Curve
precision_vals, recall_vals, _ = precision_recall_curve(y_test, y_pred_prob)
pr_auc = auc(recall_vals, precision_vals)
plt.figure(figsize=(8, 6))
plt.plot(recall_vals, precision_vals, color='blue', lw=2, label=f'PR curve (AUC = {pr_auc:.4f})')
plt.title('Precision-Recall Curve')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()

print("==========================================================")
```
